```python
%sql
WITH CTE AS (
    SELECT 
        Credit_ID,Card_Type, Credit_Name, Credit_score,
        ROW_NUMBER() OVER 
		(PARTITION BY Card_Type, Credit_Name, Credit_score ORDER BY Credit_ID) AS Rownumber
    FROM CreditScore
)select * from CTE 
```

```python
%sql
WITH CTE AS (
    SELECT 
        Credit_ID,
        ROW_NUMBER() OVER (PARTITION BY Card_Type, Credit_Name, Credit_score ORDER BY Credit_ID) AS Rownumber
    FROM CreditScore
)
DELETE FROM CreditScore
WHERE Credit_ID IN (
    SELECT Credit_ID
    FROM CTE
    WHERE Rownumber > 1
);
```
