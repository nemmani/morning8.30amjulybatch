```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, crc32, md5, sha1

# Initialize Spark session
spark = SparkSession.builder.appName("HashFunctionsDemo").getOrCreate()

# Sample data
data = [("Alice", 25), ("Bob", 30), ("Charlie", 35)]
df = spark.createDataFrame(data, ["name", "age"])

# Apply hash functions
df_hashed = df.withColumn("name1", crc32(col("name"))) \
              .withColumn("name2", md5(col("name"))) \
              .withColumn("name3", sha1(col("name")))

df_hashed.show()
#

```
