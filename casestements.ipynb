```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import when, col

# Create SparkSession
spark = SparkSession.builder \
    .appName("When-Otherwise Example") \
    .getOrCreate()

# Sample data
data = [
    (1, "visa"),
    (2, "master"),
    (3, "amex"),
    (4, "visa"),
    (5, "discover"),
    (6, "master"),
    (7, "diners")
]

columns = ["Customer_ID", "Card_Type"]

# Create DataFrame
df1 = spark.createDataFrame(data, columns)

# Apply when-otherwise logic to calculate offers
df_with_offers = df1.withColumn(
    "Offer_Percentage",
    when(col("Card_Type") == "visa", 10)
    .when(col("Card_Type") == "master", 5)
    .otherwise(1)
)

# Show the result
df_with_offers.show()
```
