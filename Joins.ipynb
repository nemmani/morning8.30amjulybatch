```python
%sql
create table orders(
    customer_id int,
    ordername varchar(50),
    orderdate date
)
```

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DateType

orders_schema = StructType([
    StructField("customer_id", IntegerType(), True),
    StructField("ordername", StringType(), True),
    StructField("orderdate", DateType(), True)
])

df1 = spark.createDataFrame([], schema=orders_schema)
display(df1)
```

```python
from datetime import datetime

data = [
    (2, 'laptop', datetime.strptime('2022-01-01', '%Y-%m-%d').date()),
    (3, 'mouse', datetime.strptime('2022-01-02', '%Y-%m-%d').date()),
    (4, 'keyboard', datetime.strptime('2022-01-03', '%Y-%m-%d').date()),
    (5, 'headphones', datetime.strptime('2022-01-04', '%Y-%m-%d').date())
]

df1 = spark.createDataFrame(data, schema=orders_schema)
display(df1)
```

```python
%sql
insert into orders values(2,'laptop','2022-01-01'),
(3,'mouse','2022-01-02'),
(4,'keyboard','2022-01-03'),
(5,'headphones','2022-01-04')
```

```python
%sql
create table customers(
    customer_id int,
    customername varchar(50),
    contactname varchar(50),
    country varchar(50)
)
    
```

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType

customers_schema = StructType([
    StructField("customer_id", IntegerType(), True),
    StructField("customername", StringType(), True),
    StructField("contactname", StringType(), True),
    StructField("country", StringType(), True)
])

df2 = spark.createDataFrame([], schema=customers_schema)
display(df2)
```

```python
data = [
    (1, 'krishna', 'maria', 'USA'),
    (2, 'ram', 'john', 'USA'),
    (3, 'shyam', 'peter', 'USA'),
    (4, 'sita', 'jack', 'USA'),
    (5, 'gita', 'mary', 'USA')
]

df2 = spark.createDataFrame(data, schema=customers_schema)
display(df2)
```

```python
%sql
insert into customers values(1,'krishna','maria','USA'),(2,'ram','john','USA'),(3,'shyam','peter','USA'),(4,'sita','jack','USA'),(5,'gita','mary','USA')
```

```python
%sql
--inner join orders, customers
select o.customer_id, o.ordername, o.orderdate, c.customername,c.contactname, c.country from orders o inner join customers c on o.customer_id = c.customer_id
```

```python
%sql
-- orders, customers
-- left join means complete left table(matching records+ unmatching records + right table matching records)
select o.customer_id, o.ordername, o.orderdate, c.customername,c.contactname, c.country from orders o left join customers c on o.customer_id = c.customer_id
```

```python
%sql
-- orders, customers
-- right join means complete right table(matching records+ unmatching records + left table matching records)
select o.customer_id, o.ordername, o.orderdate, c.customername,c.contactname, c.country from orders o right join customers c on o.customer_id = c.customer_id
```

```python
%sql
-- orders, customers
-- full join means complete both tables
select o.customer_id, o.ordername, o.orderdate, c.customername,c.contactname, c.country from orders o full join customers c on o.customer_id = c.customer_id
```
